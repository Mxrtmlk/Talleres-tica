Para este análisis, hemos seleccionado tres variables clave que combinan aspectos demográficos y de vulnerabilidad social: `sexo`, `pobreza_multi` y `area`.

La relevancia de la variable `sexo` es fundamental a nivel demográfico para estudiar la paridad y distribución de la población entre hombres y mujeres. Por otro lado, la variable `pobreza_multi` resulta esencial como indicador sensible de bienestar, ya que mide las carencias estructurales y la situación de pobreza multidimensional de los hogares. Finalmente, la variable `area` permite contextualizar territorialmente los datos al diferenciar las condiciones habitacionales entre las zonas urbanas y rurales, permitiendo visualizar cómo se distribuyen estas realidades en el país.

```python
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Gráfico para 'sexo' con anotación de frecuencias exactas
sexo_data = casen['sexo'].value_counts()
ax0 = sns.barplot(x=sexo_data.index.astype(str),
                  y=sexo_data.values, ax=axes[0], palette=['#4C72B0', '#DD8452'])
axes[0].set_title('Distribución por Sexo', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Sexo', fontsize=10)
axes[0].set_ylabel('Frecuencia', fontsize=10)
for p in ax0.patches:
    ax0.annotate(f"{int(p.get_height()):,}",
     (p.get_x() + p.get_width() / 2.,p.get_height() / 2),
                 ha='center', va='center', color='white', fontsize=11, fontweight='bold')

# Gráfico para 'pobreza_multi' con anotación de frecuencias exactas
pobreza_data = casen['pobreza_multi'].value_counts()
ax1 = sns.barplot(x=pobreza_data.index.astype(str),
                  y=pobreza_data.values, ax=axes[1], palette=['#55A868', '#C44E52'])
axes[1].set_title('Pobreza Multidimensional', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Condición', fontsize=10)
axes[1].set_ylabel('Frecuencia', fontsize=10)
for p in ax1.patches:
    ax1.annotate(f"{int(p.get_height()):,}", (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                 ha='center', va='center', color='white', fontsize=11, fontweight='bold')

# Gráfico para 'area' con anotación de frecuencias exactas
area_data = casen['area'].value_counts()
ax2 = sns.barplot(x=area_data.index.astype(str),
                  y=area_data.values, ax=axes[2], palette=['#8172B2', '#CCB974'])
axes[2].set_title('Distribución por Área', fontsize=12, fontweight='bold')
axes[2].set_xlabel('Zona', fontsize=10)
axes[2].set_ylabel('Frecuencia', fontsize=10)
for p in ax2.patches:
    ax2.annotate(f"{int(p.get_height()):,}", (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                 ha='center', va='center', color='white', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
for col in ['sexo', 'pobreza_multi', 'area']:
    print(f"--- {col} ---")
    print(casen[col].value_counts(normalize=True).round(3) * 100)
    print()
```

Al observar las distribuciones individuales, notamos lo siguiente:

*   `sexo`: Presenta una distribución equilibrada, con un ligero predominio de mujeres (~52.5%). Esto sugiere una buena representatividad base en términos de género.
*   `pobreza_multi`: Se observa un marcado desbalance; el 82.4% de los encuestados no se encuentra en situación de pobreza multidimensional, mientras que el 17.6% sí lo está. Esta disparidad es esperable en un indicador de este tipo, pero requiere atención en análisis posteriores.
*   `area`: Existe una fuerte concentración en el sector urbano (82.7%) frente al rural (17.3%), lo que refleja la distribución demográfica real del país pero marca un desbalance en la cantidad de datos disponibles por tipo de zona.

```python
for col in ['sexo', 'area', 'r3']:
    print(f"--- {col} x pobreza_multi ---")
    print(pd.crosstab(casen[col], casen['pobreza_multi'], normalize='index').round(3) * 100)
    print()
```

Algunas de las subpoblaciones interseccionales que podemos identificar son:

1. Género y pobreza multidimensional (`sexo` x `pobreza_multi`): Cruza la distribución poblacional por sexo con la condición de carencia multidimensional de los hogares, permitiendo evaluar brechas de género en la vulnerabilidad social.
2. Territorio y pobreza multidimensional (`area` x `pobreza_multi`): Analiza la concentración de la pobreza multidimensional diferenciando entre sectores urbanos y rurales.
3. Pueblos indígenas y pobreza (`r3` x `pobreza_multi`): Examina la situación de vulnerabilidad y carencias en hogares que pertenecen a alguno de los pueblos originarios reconocidos por ley.

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Gráfico de barras cruzando sexo y condición de pobreza multidimensional
sns.countplot(data=casen, x='sexo', hue='pobreza_multi', palette='Set2')

# Personalización del gráfico
plt.title('Distribución de Pobreza Multidimensional según Sexo', fontsize=14, fontweight='bold')
plt.xlabel('Sexo', fontsize=12)
plt.ylabel('Frecuencia (Observaciones)', fontsize=12)
plt.legend(title='Condición de Pobreza', labels=['Fuera de Pobreza', 'En Pobreza Multidimensional'])

plt.tight_layout()
plt.show()
```

## ÍTEM 6: Desbalance de clases

Dentro de las variables demográficas identificadas o de las que establecieron como importantes, ¿existe algún desbalance de clases? A qué creen que se puede deber, expliquen las consecuencias que esto pudiese tener al momento de usar los datos en practicas como entrenamiento de modelos de machine learning. (Si es que no identifican desbalance de todas formas, expliquen cuáles son los riesgos de ese fenómeno en el contexto de su base de datos o campo de investigación correspondiente a los datos)

- En su respuesta citen/parafraseen al menos dos artículos sobre el impacto del desbalance de clases y/o la falta de representatividad de ciertas poblaciones (variables demográficas) en bases de datos y su uso posterior.

Al analizar las variables demográficas y de bienestar seleccionadas, identificamos un claro desbalance de clases en dos de los tres indicadores principales. Mientras que el `sexo` se mantiene relativamente equilibrado (~52.5% mujeres vs ~47.5% hombres), la variable de `pobreza_multi` presenta un desbalance significativo, con aproximadamente un 82% de la muestra en categoría 'No pobre' frente a un 18% en 'Pobre'. De igual forma, el `area` muestra que un 83% de los datos provienen de zonas urbanas.

Este desbalance es crucial para el futuro entrenamiento de modelos de Machine Learning. Como señalan Suresh y Guttag (2021) en *"A Framework for Understanding Sources of Harm"*, si un grupo tiene una representatividad mucho mayor, el modelo puede aprender a minimizar el error global ignorando los patrones específicos de las minorías (subpoblaciones subrepresentadas). Además, el "Aggregation Bias" advierte que un modelo con buen rendimiento promedio podría fallar sistemáticamente en grupos específicos, como los hogares rurales o en situación de pobreza.

Finalmente, Selbst et al. (2019) recalcan que, al ser datos usados para la asignación de beneficios estatales, el desbalance de clases no es solo un problema matemático, sino sociotécnico. Un modelo sesgado por la falta de representatividad rural o indígena podría perpetuar inequidades estructurales al automatizar decisiones de política pública basándose en datos que no capturan adecuadamente la realidad de los grupos vulnerables.
